## Task 2: Read File in Different Ways

- Using the same sales_data.txt:

1. Read the entire file using .read() and print it.
2. Read the first line using .readline().
3. Read all lines using .readlines() and convert them into a list of integers.

```python
# from the previous task

sales = [1200, 450, 980, 1500, 3000]

with open('sales_data.txt','w') as file:
    for amount in sales:
        file.write(str(amount) + '\n')
```

```python
# reading the entire file using .read() method

with open('sales_data.txt','r') as file:
    contents = file.read()
    print(contents)
```

```python
# reading the first line of the file using .readline() method

with open('sales_data.txt','r') as file:
    first_line = file.readline()
    print(first_line)
```

```python
# reading all the lines of the file and saving them as a list of integers

with open('sales_data.txt','r') as file:
    lines = file.readlines()
    sales_list = [int(line.strip()) for line in lines]
    print(sales_list)
```
