## Task 4: Generate Summary Report from File

1. Read all sales values from sales_data.txt.
2. Convert them into integers.
3. Calculate and print:
    - Total Sales
    - Highest Sale
    - Lowest Sale
    - Average Sale

```python
# Reading all the file values amd converting them in integers

with open('sales_data.txt', 'r') as file:
    amount = file.readlines()
    amount_ints = [int(line.strip()) for line in amount]
    print(amount_ints)
```

```python
# Calculation 

total_sales = sum(amount_ints)
highest_sale = max(amount_ints)
loest_sale = min(amount_ints)
average_sale = total_sales / len(amount_ints)
```

```python
# printing the results of the calculations

print("Total Sales:", total_sales)
print("Highest Sale:", highest_sale)
print("Lowest Sale:", loest_sale)
print("Average Sale:", average_sale)
```
