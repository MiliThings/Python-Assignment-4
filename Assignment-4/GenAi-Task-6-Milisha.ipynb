## Task 6: Read Files Safely(Error Handling Inside File Handling Only)

- You must not use exceptions beyond file-related safeguards here.

1. Ask the user for a filename to open.
2. If the file exists, read and print it.
3. If it does not exist, print: "File not found. Please check the filename."

```python
#importing os module to check if the file exists
import os

#taking input from the user for the filename
filename = input("Enter the filename to open: ")

#checking if the file exists
if os.path.exists(filename):
    with open(filename, 'r') as file:
        contents = file.read()
        print(contents)

# printing error if the file doesnt exist
else:
    print(f"Error: The file '{filename}' does not exist.")
```
