## Task 7: Mini Project - Export Discounted Prices

1. Create a dictionary:

python
prices = {
    "Mouse": 500,
    "Keyboard": 800,
    "Monitor": 7000,
    "Pendrive": 400,
    "Camera": 5000
}

2. Ask the user for a discount percentage. 
3. Write discounted prices into discount_report.txt using: Product | Original Price | Discounted Price After writing, read the file and print it to the terminal.

- Extra (optional): Write a summary at the bottom of the file: Total Items: X, Average Discounted Price: Y

```python
# creating a dictionary
prices = {"Mouse":500,
          "Keyboard":800,
          "Monitor":7000,
          "Pendrive":400,
          "Camera":5000}
```

```python
# asking user for a discount

discount_percentage = float(input("Enter the discount percentage: "))
# applying the discount to each item
with open("discount_report.txt","w") as file:
    discount_value = []
    for product, original_price in prices.items():
        discounted_price = original_price - (original_price * (discount_percentage/100))
        discount_value.append(discounted_price)
        file.write(f"{product} | {original_price} | {discounted_price}\n")

    total_items = len(prices)
    total_average_discounted_price = sum(discount_value) / total_items
    file.write(f"\nTotal Items: {total_items}\n")
    file.write(f"Average Discounted Price: {total_average_discounted_price}\n")
    
```

```python
print(f"Applying a discount of {discount_percentage}% to all items.")

with open("discount_report.txt","r") as file:
    report_contents = file.read()
    print(report_contents)
```
