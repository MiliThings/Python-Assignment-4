## Task 3: Append New Sales

1. Append these new sales to the same file: 5000,2500,1700
2. After Appending, reope and print the entire updated file.

- Extra: Print the total number of lines after appending

```python
# Appending the sales data to the existing file
with open('sales_data.txt', 'a') as file:
    new_sales = [2000, 2500, 3000]
    for amount in new_sales:
        file.write(str(amount) + '\n')
```

```python
# reopening the file to read the contents

with open('sales_data.txt', 'r') as file:
    contents = file.read()
    print(contents)
```

```python
# calculating the total number of lines after appending the new sales data
with open("sales_data.txt", "r") as file:
    lines = file.readlines()
    print("Total number of lines:", lines, len(lines))
```
