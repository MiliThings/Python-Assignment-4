## Task 5: Create Product info File

1. Ask the user for 3 product names & their prices.
2. Write them into a new file products.txt in this format: ProductName | Price
3. Read the file and print each line with proper formatting.

```python
# Asking users for 3 product names and their prices

with open("products.txt","w") as file:
    for i in range(3):
        product_name = input(f"Enter the name of product {i+1}: ")
        product_price = input(f"Enter the price of {product_name}: ")
        file.write(f"{product_name} | {product_price}\n")
```

```python
# Reading the products and their prices from the file
with open("products.txt","r") as file:
    contents = file.readlines()
    print("\nProducts and Prices:")
    for line in contents:
        print(line.strip())
```
